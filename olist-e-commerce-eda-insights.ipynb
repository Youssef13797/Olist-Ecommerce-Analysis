## 🔗 Merging Datasets
### The Olist dataset is built like a relational database. To get a comprehensive view of our business, we need to join these tables into a single, unified DataFrame. 
### We will merge the orders table with customers, items, products (translated to English), sellers, payments, and reviews using their respective primary keys.

```python
payments_summary = payments.groupby('order_id', as_index=False)['payment_value'].sum()

reviews_summary = reviews.groupby('order_id', as_index=False)['review_score'].mean()

products_en = pd.merge(products, translation, on='product_category_name', how='left')
```

```python
df = pd.merge(orders_dataset, customers, on='customer_id', how='left')

df = pd.merge(df, items, on='order_id', how='left')

df = pd.merge(df, products_en, on='product_id', how='left')

df = pd.merge(df, sellers, on='seller_id', how='left')

df = pd.merge(df, payments_summary, on='order_id', how='left')

df = pd.merge(df, reviews_summary, on='order_id', how='left')
df
```

## 🧹 1. Data Cleaning & Preprocessing
Before analyzing the data, we must ensure its quality. In this section we will:
* Filter only completed/delivered orders.
* Handle missing values intelligently without losing crucial data.
* Translate and standardize Brazilian State abbreviations to full names for better mapping.

```python
date_columns = [
    'order_purchase_timestamp', 
    'order_approved_at', 
    'order_delivered_carrier_date', 
    'order_delivered_customer_date', 
    'order_estimated_delivery_date',
    'shipping_limit_date'
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col])

print(df[date_columns].dtypes)
```

```python
df.isnull().sum()
```

```python
df['order_status'].value_counts()
```

```python
px.pie(df, names='order_status', title='Distribution of Order Statuses').show()
```

```python
df = df.dropna(subset=['product_id'])
df['product_category_name_english'] = df['product_category_name_english'].fillna('Unknown')
df['review_score'] = df['review_score'].fillna(0)
```

```python
df.isnull().sum()
```

## 🛠️ Handling Missing Values (The Business Logic)

### Data cleaning is not just about removing empty rows; it is about distinguishing between **data errors** and **business facts**. We adopted a strategic approach to preserve the context of our operations:

### 1. **The "Natural" Nulls (Operational Insights):** We kept the missing values in `order_delivered_customer_date` (2,454 rows) and `order_delivered_carrier_date` (1,194 rows). These are **not errors**. They represent orders that were either `canceled` or `unavailable` and therefore never reached the delivery stage. Dropping these would bias our analysis and hide critical operational bottlenecks.

### 2. **Minor Administrative Gaps:** The 15 missing values in `order_approved_at` are negligible and do not impact our macro-level analysis, so they were left as is to maintain data integrity.

### 3. **Critical Data Filtering:** We strictly dropped rows missing essential features for financial and physical calculations (e.g., `payment_value` or `product_weight_g`), ensuring that our revenue and logistics insights remain accurate.

### 4. **Categorical Fill:** Missing product categories were labeled as `'Unknown'` rather than dropping the entire record. This allows us to keep the transaction in our sales volume reports while acknowledging the missing taxonomy.

```python
cols_to_drop = ['product_weight_g', 'payment_value']
df = df.dropna(subset=cols_to_drop)

cols_to_zero = ['product_name_lenght', 'product_description_lenght', 'product_photos_qty']
for col in cols_to_zero:
    df[col] = df[col].fillna(0)
    
df['product_category_name'] = df['product_category_name'].fillna('Unknown')
```

## 📊 2. Exploratory Data Analysis (EDA)

### 📈 2.1 Sales Trend Over Time
Let's look at the big picture. How have order volumes changed since the company started?

```python
filtered_df = df[(df['order_purchase_timestamp'] >= '2017-01-01') & 
                 (df['order_purchase_timestamp'] <= '2018-08-31')]

filtered_df['year_month'] = filtered_df['order_purchase_timestamp'].dt.to_period('M').astype(str)
monthly_orders = filtered_df.groupby('year_month').size().reset_index(name='order_count')
monthly_orders = monthly_orders.sort_values('year_month')

fig = px.line(
    monthly_orders, 
    x='year_month', 
    y='order_count', 
    title='📉 Monthly Orders Trend (Jan 2017 - Aug 2018)',
    markers=True,
    labels={'year_month': 'Month', 'order_count': 'Total Orders'}
)

fig.update_traces(line_color='#1f77b4', line_width=3)
fig.show()
```

**📌 Insight:**
*We can observe a continuous growth in orders over time, with significant peaks likely related to holiday seasons or massive discount events like Black Friday.*

```python
top_cities = df['customer_city'].value_counts().head(10).reset_index()
top_cities.columns = ['customer_city', 'order_count']

fig = px.bar(
    top_cities, 
    x='customer_city', 
    y='order_count',
    title='🏙️ Top 10 Cities by Orders',
    text='order_count',
    color='order_count',
    color_continuous_scale='Blues'
)

fig.update_traces(textposition='outside')
fig.show()
```

**📌 Insight:**
*Unsurprisingly, **São Paulo** completely dominates the customer base, followed by Rio de Janeiro. This geographic concentration means that establishing distribution centers or optimizing supply chains in the Southeast region of Brazil will yield the highest return on investment (ROI) and significantly reduce delivery times.*

```python
delivered_orders = df[df['order_status'] == 'delivered']
top_categories_revenue = delivered_orders.groupby('product_category_name_english')['price'].sum().sort_values(ascending=False).head(10).reset_index()

top_categories_revenue = df.groupby('product_category_name_english')['price'].sum().sort_values(ascending=False).head(10).reset_index()

fig = px.bar(
    top_categories_revenue, 
    x='price',
    y='product_category_name_english', 
    orientation='h',
    title='🏆 Top 10 Categories by Revenue',
    text_auto='.2s',
    color='price',
    color_continuous_scale='Viridis'
)

fig.update_layout(yaxis={'categoryorder':'total ascending'})
fig.show()
```

**📌 Insight:**
*While some items might sell in high quantities, the **Health & Beauty** and **Watches/Gifts** categories are the true cash cows, generating the most revenue. Marketing budgets should heavily target these high-margin categories.*

```python
failed_orders = df[df['order_status'].isin(['canceled', 'unavailable'])]

top_failed_categories = failed_orders.groupby('product_category_name_english').size().reset_index(name='failed_count')
top_failed_categories = top_failed_categories.sort_values('failed_count', ascending=False).head(10)

fig_failed = px.bar(
    top_failed_categories,
    x='failed_count',
    y='product_category_name_english',
    orientation='h',
    title='🚨 Top 10 Categories by Failed Orders (Canceled & Unavailable)',
    text='failed_count',
    color='failed_count',
    color_continuous_scale='Reds'
)

fig_failed.update_layout(yaxis={'categoryorder':'total ascending'})
fig_failed.show()
```

**📌 Insight:**
*By keeping the non-delivered orders, we discovered critical operational bottlenecks. The categories shown in red above suffer from the highest cancellation and unavailability rates. The business needs to immediately investigate the inventory and supply chain partners handling these specific categories to reduce revenue leakage.*

```python
review_dist = df['review_score'].value_counts().reset_index()
review_dist.columns = ['review_score', 'count']

review_dist['review_score'] = review_dist['review_score'].astype(str) + ' Stars'

fig = px.pie(
    review_dist, 
    values='count', 
    names='review_score',
    title='⭐ Review Score Distribution',
    hole=0.4, 
    color_discrete_sequence=px.colors.qualitative.Pastel
)

fig.update_traces(textposition='inside', textinfo='percent+label')
fig.show()
```

**📌 Insight:**
*The majority of customers are highly satisfied, with nearly **60% giving a 5-star rating**. However, there is a noticeable chunk (around 11%) of 1-star reviews. As a next step for the business, investigating these 1-star reviews (e.g., correlating them with late delivery times) would be crucial to improve customer retention.*

### ⏰ 2.4 Peak Shopping Hours (Heatmap)
When do Brazilians prefer to shop online? By extracting the day and hour from the timestamp, we can identify the busiest times for the platform.

```python
df['order_hour'] = df['order_purchase_timestamp'].dt.hour
df['order_day'] = df['order_purchase_timestamp'].dt.day_name()

days_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

heatmap_data = df.groupby(['order_day', 'order_hour']).size().reset_index(name='orders_count')

heatmap_matrix = heatmap_data.pivot(index='order_day', columns='order_hour', values='orders_count')
heatmap_matrix = heatmap_matrix.reindex(days_order)

fig = px.imshow(
    heatmap_matrix,
    labels=dict(x="Hour", y="Day", color="Number of orders"),
    x=heatmap_matrix.columns,
    y=heatmap_matrix.index,
    title='🔥 Heatmap of Peak Order Times',
    color_continuous_scale='YlOrRd'
)

fig.show()
```

**📌 Insight:**
*The heatmap clearly shows that peak shopping activity occurs during **weekdays (Monday to Friday) between 10:00 AM and 4:00 PM**. This suggests that many customers browse and buy during their work hours or lunch breaks! The marketing team should schedule their heavy ad campaigns during these exact hours.*
